使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。
n = int(input())
for i in range(1, n + 1):
  row = []
  for j in range(1, n + 1):
    # 計算相乘結果
    product = i * j
    # 使用 f-string 進行格式化與欄位對齊（例如確保運算式與結果整齊排列）
    # 這裡採用常見的格式：j * i = 乘積，並設定固定寬度對齊
    expression = f"{j} * {i} = {product:2d}"
    row.append(expression)
  print("  ".join(row))

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。
n = int(input())
i = 1
while i <= n:
  j = 1
  row = []
  while j <= n:
    product = i * j
    # 使用 f-string 的 :b 將數字轉為二進位，並透過欄位寬度（例如 >3b 或 >6b）進行對齊
    expression = f"{j:>3b} * {i:>3b} = {product:>6b}"
    row.append(expression)
    j += 1
  print("  ".join(row))
  i += 1

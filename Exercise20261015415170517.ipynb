習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png
import random

# 1. 產生一個介於 0 到 1 之間的小數 r
r = random.random()
print(f"產生的隨機小數 r = {r}")

# 2. 使用 while 迴圈計算前二十位二進位表示
binary_digits = []
temp = r
count = 0

while count < 20:
    temp *= 2
    bit = int(temp)  # 取整數部分作為二進位的一位
    binary_digits.append(str(bit))
    temp -= bit      # 保留小數部分繼續循環
    count += 1

# 3. 組合輸出結果
binary_result = "".join(binary_digits)
print(f"前二十位二進位表示: 0.{binary_result}")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。
def int_to_chinese(n: int) -> str:
    # 定義中文數字與基礎單位
    num_chars = ["零", "壹", "貳", "參", "肆", "伍", "陸", "柒", "捌", "玖"]
    pos_units = ["", "拾", "佰", "仟"]
    
    # 定義大單位（對應 10^4 的冪次）
    # 個, 萬, 億, 兆, 京, 垓, 秭, 穰, 溝, 澗, 正, 載, 極, 恆河沙, 阿僧祇, 那由他, 不可思議, 無量大數
    big_units = [
        "", "萬", "億", "兆", "京", "垓", "秭", "穰", "溝", "澗", 
        "正", "載", "極", "恆河沙", "阿僧祇", "那由他", "不可思議", "無量大數"
    ]
    
    if n == 0:
        return "零"
    
    str_n = str(n)
    length = len(str_n)
    
    # 將數字每 4 位分組（從右到左）
    groups = []
    for i in range(length, 0, -4):
        groups.append(str_n[max(0, i - 4):i])
    
    res_parts = []
    zero_flag = False
    
    for g_idx, group in enumerate(groups):
        g_val = int(group)
        if g_val == 0:
            # 整個群組皆為 0 時的處理
            if not zero_flag and res_parts and g_idx > 0:
                res_parts.append("零")
                zero_flag = True
            continue
        
        # 轉換四位數以內的數字
        group_str = ""
        g_len = len(group)
        for i, ch in enumerate(group):
            digit = int(ch)
            position = g_len - 1 - i
            if digit != 0:
                if zero_flag:
                    group_str += "零"
                    zero_flag = False
                group_str += num_chars[digit] + pos_units[position]
            else:
                if not zero_flag and group_str and i < g_len - 1:
                    zero_flag = True
        
        # 加上對應的大單位
        group_str += big_units[g_idx]
        res_parts.insert(0, group_str)
        zero_flag = False

    return "".join(res_parts)

# 測試範例
n = 31415926
print(f"輸入: {n}")
print(f"輸出: {int_to_chinese(n)}")[cite: 1]

習題三：請勿使用for、range，而是利用iterable、iter()、next()、while、StopIteration來設計九九乘法表。
def print_multiplication_table():
    # 利用 iter() 將範圍轉換成迭代器，不使用 for 與 range 關鍵字控制
    rows = iter(range(1, 10))
    
    try:
        while True:
            i = next(rows)
            cols = iter(range(1, 10))
            line_results = []
            try:
                while True:
                    j = next(cols)
                    line_results.append(f"{i}x{j}={i*j:2d}")
            except StopIteration:
                pass
            print("  ".join(line_results))
    except StopIteration:
        pass

# 執行九九乘法表
print_multiplication_table()

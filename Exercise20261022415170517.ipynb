### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

值：{"multiplier": 1.1, "bonus": 5}

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
*（乘法運算子）：在 score * multiplier 中代表數值相乘（計算加權分數）。若在其他情境（如字串或列表）則代表重複，或在函數定義/呼叫時代表打包/解包。

**（次方運算子）：在 score ** 2 中代表「次方」（平方，即 score 的 2 次方）。若在函數定義/呼叫時則代表關鍵字參數打包/解包。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？
print() 只是將資料印在螢幕上給使用者看，函數本身的回傳值是 None，其他程式碼無法再利用這些計算結果。

return 會將計算結果回傳給呼叫端（存入變數 result），讓程式後續可以針對這些數據進行分析、儲存、傳送或進一步運算。

請思考並回答：

1. 原本的 `*scores` 與 `scores_list` 有什麼不同？
*scores 是位置參數打包，接收獨立的多個數值並組合成 tuple。

scores_list 是關鍵字參數，直接接收一個已存在的串列（list）。

因為原本的函數定義是 def analyze_scores(*scores, **options):：

如果傳入 scores_list=[60, 70, 80]，這個參數會被 **options 收集進去，變成 options 字典的一個鍵值對 {"scores_list": [60, 70, 80]}。

然而，程式內部的 for score in scores: 是去走訪 scores（此時 scores 會是空的 ())，而不是去走訪 scores_list，導致無法正確讀取分數。

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

星號的用途
乘法運算：用於數值計算（例如：score * multiplier）。

函數定義端（打包）：*args 將多個獨立的位置參數打包成一個 tuple。

函數呼叫端（解包）：*scores 將可迭代物件（如串列、元組）拆解成獨立的位置參數傳入。
def show_items(*args):   # 定義端：打包成 tuple
    print(args)

my_list = [1, 2, 3]
show_items(*my_list)     # 呼叫端：解包為獨立參數 (輸出: (1, 2, 3))
result = 5 * 2           # 乘法用途 (輸出: 10)

** 雙星號的用途
次方運算：用於計算冪次（例如：score ** 2 代表平方）。

函數定義端（打包）：**kwargs 將多個未命名的關鍵字參數打包成一個 dict。

函數呼叫端（解包）：**settings 將字典的鍵值對拆解成獨立的關鍵字參數傳入。
def config_settings(**kwargs):  # 定義端：打包成 dict
    print(kwargs)

d = {"multiplier": 2, "bonus": 10}
config_settings(**d)            # 呼叫端：解包為關鍵字參數
power = 3 ** 2                  # 次方用途 (輸出: 9)

return 的用途
終止函數並回傳結果：將函數內部的計算成果交給呼叫端，同時結束該函數的執行。

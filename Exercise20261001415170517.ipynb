1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
十進位小數轉換為二進位小數時，採用「連乘 2 取整數部分」的方法：$0.1 \times 2 = 0.2$ $\rightarrow$ 整數部分為 0，小數留 $0.2$$0.2 \times 2 = 0.4$ $\rightarrow$ 整數部分為 0，小數留 $0.4$$0.4 \times 2 = 0.8$ $\rightarrow$ 整數部分為 0，小數留 $0.8$$0.8 \times 2 = 1.6$ $\rightarrow$ 整數部分為 1，小數留 $0.6$$0.6 \times 2 = 1.2$ $\rightarrow$ 整數部分為 1，小數留 $0.2$$0.2 \times 2 = 0.4$ $\rightarrow$ 整數部分為 0，小數留 $0.4$（開始進入循環）計算結果：十進位的 0.1 轉換為二進位後會變成無窮循環小數：$$0.1_{10} = 0.00011001100110011..._2 = 0.0\overline{0011}_2$$觀察到的現象：十進位中看似簡單的有限小數 0.1，在二進位中卻是無窮循環小數（因為 10 的質因數包含 5，而二進位只能精確表示分母為 $2^n$ 的小數）。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
因為電腦記憶體有限，無法儲存無限多位的二進位小數，Python 主要採用以下機制來處理：

IEEE 754 雙精度浮點數標準（Double Precision）：

Python 的 float 型別佔用 64 位元（bit）：1 位元為符號位（Sign）、11 位元為指數（Exponent）、52 位元為尾數/有效數字（Fraction/Mantissa）。

尾數最多儲存 52 位，多餘的位元必須被截斷或捨入。

捨入機制：「四捨六入五成雙」（Banker's Rounding / Round to Nearest, Ties to Even）：

當捨去部分剛好等於半數（例如二進位尾數的 .1000...）時，會捨入到最近的偶數。

這種捨入規則能有效減少多次計算後產生的正向或負向累積偏誤。

因此，二進位的 0.1 在記憶體中會被儲存為一個極度接近但不完全等於 0.1 的估算值

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
  1. 金融領域（Financial Applications）—— 不是好方法（應避免使用 float）
原因：金融領域對金額要求絕對精準（「一分錢都不能少」）。float 產生的累積微小誤差累積到大規模交易或複利計算時，會導致帳務不符。

最佳做法：使用 decimal.Decimal（基於十進位算術規律）或直接將單位轉換為最小貨幣單位（如「分」）改用整數（int）計算。

  2. 科學計算與工程領域（Scientific Computing）—— 是標準且合適的方法
原因：科學計算通常涉及矩陣運算與大規模資料處理，更在乎計算速度與記憶體效能。CPU 硬體本身對 IEEE 754 浮點數有原生優化支援（如 GPU 加速、AVX 指令集）。

最佳做法：直接使用原生 float 或 NumPy 的 float64 / float32，只要確保數值穩定性（Numerical Stability），並在比較相等時使用相對誤差Tolerance（如 math.isclose() 或 np.isclose()）即可。
